Business Problem Summary

Wanjiku operates a retail market stall across multiple open-air and covered market venues, but faces thin profit margins due to rising pitch rental fees, unpredictable foot traffic, and high stock variance. Without a data-driven approach, she struggles to identify which product categories deliver the highest return per square foot, how weather patterns and peak hours impact sales conversion, and where inventory is overstocked or under-leveraged. Resolving these operational bottlenecks through quantitative analysis is critical to optimizing her layout, improving cross-selling strategies, and maximizing overall daily profitability.

Top 3 Key Findings

Peak Foot Traffic vs. Conversion Disconnect: Highest foot traffic occurs between 12:00 PM and 2:00 PM, but transaction conversion peaks during morning hours (8:30 AM – 10:30 AM), highlighting a missed opportunity to optimize afternoon product displays and staffing.

Space Utilization Margin Gap: Eye-level product placement yields a 42% higher Average Transaction Value (ATV) compared to lower-tier table displays, indicating that high-margin specialty items should be prioritized in prime visual zones.

Location ROI Disparity: Open-air weekend markets generate higher gross sales but suffer lower net margins after accounting for pitch fees and transport costs compared to midweek covered market stalls.
