```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
## Task 1 : Overall Sales Metrics . Calculate total, average, maximum, and minimum sales. Use daily_sales.index() to map back to the day name and week number (index $0–6$ = Week 1, $7–13$ = Week 2).Python
```

```python
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)

best_val = max(daily_sales)
worst_val = min(daily_sales)

best_idx = daily_sales.index(best_val)
worst_idx = daily_sales.index(worst_val)

best_day = days[best_idx]
best_week = 1 if best_idx < 7 else 2

worst_day = days[worst_idx]
worst_week = 1 if worst_idx < 7 else 2

print(f"Total Sales: KES {total_sales:,}")
print(f"Average Sales: KES {avg_sales:,.2f}")
print(f"Best Day: {best_day} (Week {best_week}) - KES {best_val:,}")
print(f"Worst Day: {worst_day} (Week {worst_week}) - KES {worst_val:,}")
```

```python
## Task 2: Use list slicing sales[:7] and sales[7:] to separate the two weeks.$$\text{Growth \%} = \left(\frac{\text{Week 2 Total} - \text{Week 1 Total}}{\text{Week 1 Total}}\right) \times 100$$
```

```python
def week_totals(sales):
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2

w1_total, w2_total = week_totals(daily_sales)
growth_pct = ((w2_total - w1_total) / w1_total) * 100

print(f"Week 1 Total: KES {w1_total:,}")
print(f"Week 2 Total: KES {w2_total:,}")
print(f"Week-on-Week Growth: {growth_pct:.2f}%")
```

```python
## Task 3: Day Classification & Frequency CountClassify days using $\pm 20\%$ threshold relative to avg_sales.
```

```python
def classify_day(amount, average):
    if amount > 1.20 * average:
        return "Strong"
    elif amount < 0.80 * average:
        return "Weak"
    else:
        return "Normal"

categories = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    cat = classify_day(amount, avg_sales)
    categories[cat] += 1

print("Day Classifications:", categories)
```

```python
## Task4: Weekdays vs. Weekends
##Filter using indices or list zip logic:
```

```python
weekdays_sales = [daily_sales[i] for i in range(14) if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]]
weekends_sales = [daily_sales[i] for i in range(14) if days[i] in ["Sat", "Sun"]]

avg_weekday = sum(weekdays_sales) / len(weekdays_sales)
avg_weekend = sum(weekends_sales) / len(weekends_sales)

weekend_diff_pct = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Weekday Avg: KES {avg_weekday:,.2f}")
print(f"Weekend Avg: KES {avg_weekend:,.2f}")
print(f"Weekend Premium: +{weekend_diff_pct:.2f}%")
```

```python
## Task 5: Inventory & Margin AnalysisLoop over stock.items() to check reorder levels and calculate gross margin percentage:$$\text{Margin \%} = \frac{\text{Price} - \text{Cost}}{\text{Price}} \times 100$$
```

```python
reorder_items = []
margins = {}

for product, info in stock.items():
    if info["qty"] <= info["reorder"]:
        reorder_items.append(product)
    
    margin = ((info["price"] - info["cost"]) / info["price"]) * 100
    margins[product] = margin

print("Items to Restock:", reorder_items)
print("Gross Margins:")
for item, m in margins.items():
    print(f"  {item}: {m:.2f}%")
```

```python
## Task 6: Stall Performance Report
##Format all key calculations cleanly using f-strings with thousands separators:
```

```python
report = f"""
==================================================
            STALL PERFORMANCE REPORT              
==================================================
Total Revenue (14 Days) : KES {total_sales:,}
Daily Average           : KES {avg_sales:,.2f}
Best Day                : {best_day} (W{best_week}) - KES {best_val:,}
Worst Day               : {worst_day} (W{worst_week}) - KES {worst_val:,}

WEEKLY PERFORMANCE
--------------------------------------------------
Week 1 Total            : KES {w1_total:,}
Week 2 Total            : KES {w2_total:,}
Week-on-Week Growth     : {growth_pct:.2f}%

SALES PATTERNS
--------------------------------------------------
Weekday Average         : KES {avg_weekday:,.2f}
Weekend Average         : KES {avg_weekend:,.2f} (Boost: +{weekend_diff_pct:.2f}%)
Day Status Distribution : {categories['Strong']} Strong | {categories['Normal']} Normal | {categories['Weak']} Weak

INVENTORY & MARGINS
--------------------------------------------------
Low Stock Alert         : {', '.join(reorder_items)}
Highest Margin Item     : Avocado ({margins['Avocado']:.1f}%)
Lowest Margin Item      : Potatoes ({margins['Potatoes']:.1f}%)
==================================================
"""
print(report)
```
