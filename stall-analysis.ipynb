```python

# ## Task 1: Total Weekly Revenue and Sales Volume
# **Approach:** Iterate through the daily sales dataset to calculate the aggregate revenue and total quantity of items sold across the entire 7-day period.

# %%
dataset = [
    {"day": "Monday", "revenue": 12500, "units_sold": 45, "foot_traffic": 320},
    {"day": "Tuesday", "revenue": 9800, "units_sold": 38, "foot_traffic": 280},
    {"day": "Wednesday", "revenue": 11200, "units_sold": 41, "foot_traffic": 300},
    {"day": "Thursday", "revenue": 10500, "units_sold": 39, "foot_traffic": 290},
    {"day": "Friday", "revenue": 18200, "units_sold": 68, "foot_traffic": 450},
    {"day": "Saturday", "revenue": 24500, "units_sold": 92, "foot_traffic": 610},
    {"day": "Sunday", "revenue": 21800, "units_sold": 81, "foot_traffic": 540}
]
```

```python
def week_totals(data):
    """Calculates total weekly revenue, total units sold, and overall foot traffic."""
    total_rev = sum(item["revenue"] for item in data)
    total_units = sum(item["units_sold"] for item in data)
    total_traffic = sum(item["foot_traffic"] for item in data)
    return total_rev, total_units, total_traffic

total_revenue, total_units, total_traffic = week_totals(dataset)
```

```python
# ## Task 2: Average Daily Performance Metrics
# **Approach:** Divide the total weekly aggregates by the total number of operational market days to evaluate average daily baseline performance.

# %%
num_days = len(dataset)
avg_daily_revenue = total_revenue / num_days
avg_daily_units = total_units / num_days
avg_daily_traffic = total_traffic / num_days
```

```python
# ## Task 3: Day Classification Based on Revenue
# **Approach:** Define a function that classifies each day's performance based on revenue thresholds to identify high-performing and low-performing market days.

# %%
def classify_day(revenue):
    """Classifies market day performance based on revenue thresholds."""
    if revenue >= 20000:
        return "High Performing"
    elif revenue >= 12000:
        return "Moderate Performing"
    else:
        return "Low Performing"
```

```python
classified_days = [
    {"day": item["day"], "revenue": item["revenue"], "category": classify_day(item["revenue"])}
    for item in dataset
]
```

```python
## Task 4: Conversion Rate and Average Transaction Value (ATV)
# **Approach:** Calculate the conversion rate (percentage of foot traffic resulting in sales) and ATV (average revenue generated per transaction unit) for each day.

# %%
daily_analytics = []
for item in dataset:
    conversion_rate = (item["units_sold"] / item["foot_traffic"]) * 100
    atv = item["revenue"] / item["units_sold"] if item["units_sold"] > 0 else 0
    daily_analytics.append({
        "day": item["day"],
        "revenue": item["revenue"],
        "conversion_rate": conversion_rate,
        "atv": atv
    })
```

```python
## Task 5: Sorting Days by Revenue Performance
# **Approach:** Implement a custom reordering function using a simple sorting technique to rank operational days from highest revenue to lowest.

# %%
def reorder_by_revenue(data):
    """Sorts the list of daily records in descending order of revenue without external libraries."""
    return sorted(data, key=lambda x: x["revenue"], reverse=True)

sorted_dataset = reorder_by_revenue(dataset)
```

```python
# ## Task 6: Stall Performance Report Generation
# **Approach:** Format and print a clean, executive summary text report displaying aggregate, daily, and ranked metrics with KES currency formatting.

# %%
print("=" * 55)
print("             STALL PERFORMANCE REPORT             ")
print("=" * 55)
print(f"Total Revenue:       KES {total_revenue:,.2f}")
print(f"Total Units Sold:    {total_units:,} units")
print(f"Total Foot Traffic:  {total_traffic:,} visitors")
print("-" * 55)
print(f"Average Daily Rev:   KES {avg_daily_revenue:,.2f}")
print(f"Average Daily Units: {avg_daily_units:.1f} units")
print(f"Average Daily Traffic:{avg_daily_traffic:.1f} visitors")
print("=" * 55)
print(f"{'Day':<12} | {'Revenue (KES)':<14} | {'Category':<18}")
print("-" * 55)
for item in sorted_dataset:
    cat = classify_day(item['revenue'])
    print(f"{item['day']:<12} | KES {item['revenue']:<10,} | {cat:<18}")
print("=" * 55)
```
