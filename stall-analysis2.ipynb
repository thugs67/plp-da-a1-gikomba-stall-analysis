```python
## Task 1: Total Weekly Revenue and Sales Volume (14-Day Dataset)
# **Approach:** Replace synthetic data with the 14-day sales list and stock dictionary. Use list slicing (`data[:7]` for Week 1 and `data[7:]` for Week 2) within `week_totals()` to evaluate each week independently without external dependencies.

# %%
# 14-Day Sales Dataset
dataset = [
    {"day": "Monday (W1)", "revenue": 12500, "units_sold": 45, "foot_traffic": 320},
    {"day": "Tuesday (W1)", "revenue": 9800, "units_sold": 38, "foot_traffic": 280},
    {"day": "Wednesday (W1)", "revenue": 11200, "units_sold": 41, "foot_traffic": 300},
    {"day": "Thursday (W1)", "revenue": 10500, "units_sold": 39, "foot_traffic": 290},
    {"day": "Friday (W1)", "revenue": 18200, "units_sold": 68, "foot_traffic": 450},
    {"day": "Saturday (W1)", "revenue": 24500, "units_sold": 92, "foot_traffic": 610},
    {"day": "Sunday (W1)", "revenue": 21800, "units_sold": 81, "foot_traffic": 540},
    {"day": "Monday (W2)", "revenue": 13100, "units_sold": 48, "foot_traffic": 330},
    {"day": "Tuesday (W2)", "revenue": 10200, "units_sold": 40, "foot_traffic": 290},
    {"day": "Wednesday (W2)", "revenue": 11800, "units_sold": 44, "foot_traffic": 310},
    {"day": "Thursday (W2)", "revenue": 11000, "units_sold": 41, "foot_traffic": 300},
    {"day": "Friday (W2)", "revenue": 19500, "units_sold": 72, "foot_traffic": 470},
    {"day": "Saturday (W2)", "revenue": 26000, "units_sold": 98, "foot_traffic": 640},
    {"day": "Sunday (W2)", "revenue": 23000, "units_sold": 85, "foot_traffic": 560}
]
```

```python
# Stock Inventory Dictionary
stock_inventory = {
    "Produce": {"starting_stock": 500, "reorder_point": 100},
    "Dairy": {"starting_stock": 300, "reorder_point": 60},
    "Dry Goods": {"starting_stock": 400, "reorder_point": 80}
}
```

```python
def week_totals(data):
    """Calculates revenue, units sold, and foot traffic totals for Week 1 and Week 2 using list slicing."""
    week1_data = data[:7]
    week2_data = data[7:]
    
    def calc_totals(week_slice):
        rev = sum(item["revenue"] for item in week_slice)
        units = sum(item["units_sold"] for item in week_slice)
        traffic = sum(item["foot_traffic"] for item in week_slice)
        return {"revenue": rev, "units_sold": units, "foot_traffic": traffic}
    
    return calc_totals(week1_data), calc_totals(week2_data)

w1_totals, w2_totals = week_totals(dataset)
```

```python
# ## Task 2: Average Daily Performance Metrics
# **Approach:** Compute average daily baseline performance across the full 14-day operational period.

# %%
total_revenue = sum(item["revenue"] for item in dataset)
total_units = sum(item["units_sold"] for item in dataset)
total_traffic = sum(item["foot_traffic"] for item in dataset)

num_days = len(dataset)
avg_daily_revenue = total_revenue / num_days
avg_daily_units = total_units / num_days
avg_daily_traffic = total_traffic / num_days
```

```python
## Task 3: Day Classification Based on Revenue
# **Approach:** Classify each day's performance based on revenue thresholds.

# %%
def classify_day(revenue):
    """Classifies market day performance based on revenue thresholds."""
    if revenue >= 20000:
        return "High Performing"
    elif revenue >= 12000:
        return "Moderate Performing"
    else:
        return "Low Performing"

classified_days = [
    {"day": item["day"], "revenue": item["revenue"], "category": classify_day(item["revenue"])}
    for item in dataset
]
```

```python
## Task 4: Conversion Rate and Average Transaction Value (ATV)
# **Approach:** Calculate conversion rate and ATV across the 14-day timeline.

# %%
daily_analytics = []
for item in dataset:
    conversion_rate = (item["units_sold"] / item["foot_traffic"]) * 100
    atv = item["revenue"] / item["units_sold"] if item["units_sold"] > 0 else 0
    daily_analytics.append({
        "day": item["day"],
        "revenue": item["revenue"],
        "conversion_rate": conversion_rate,
        "atv": atv
    })
```

```python
# ## Task 5: Sorting Days by Revenue Performance
# **Approach:** Sort the 14-day dataset in descending order of revenue.

# %%
def reorder_by_revenue(data):
    """Sorts the list of daily records in descending order of revenue without external libraries."""
    return sorted(data, key=lambda x: x["revenue"], reverse=True)

sorted_dataset = reorder_by_revenue(dataset)
```
